## Admission à l'hôpital et en réanimation jusqu'au 18 mars 2021 [(padlet page 7)](https://padlet.com/julia_dumont/covid19-les-graphiques-nifnaf-puilc4nj7dwj/slideshow/wish/1331645801)

```python
pf.setwhom('owid',reload=False)
pf.plot(which='cur_hosp_patients',where='FRA',what="daily",option='smooth7',when='18/03/2020:18/03/2021')
pf.plot(which='cur_icu_patients',where='FRA',when='18/03/2020:18/03/2021')
```

On comprend ici que sur la courbe diffusée, on a le nb de lits de réa d'une part, et le nombre d'admis à l'hopital (pas en réa) de l'autre.

# Divers

```python
import pandas as pd
pf.setwhom('spf',reload=False)
pf.setvis('bokeh')

spf=pf.get(which='cur_hosp',when='23/03/2020:28/03/2021',option='sumall')
spf=spf.drop(columns='geometry')
spf['where']=len(spf)*['France']
pf.setwhom('owid',reload=False)
owid=pf.get(which='cur_icu_patients',when='23/03/2020:28/03/2021',where='France')
owid=owid.drop(columns='geometry')

merge=pd.merge(spf,owid,on=['date','where'],how='outer')
pf.plot(input=merge,which=['cur_icu_patients','cur_hosp'])
```
